## 3.6 Data Structures

### Lists

A **list** stores values in a fixed order. The index starts at 0, negative indices count from the end. Lists are **mutable**.

````python
order = ["Espresso", "Croissant", "Espresso"]
print(order[0], order[-1])
print(order[0:2])
print(len(order))
````

````python
order.append("Muffin")
order.remove("Croissant")
print(order)
print("Muffin" in order, order.count("Espresso"))
````

### Tuples

A **tuple** is a fixed record, for example one receipt line. Tuples are **immutable**.

````python
line = ("Espresso", 1.10, 3)
item, price, quantity = line     # unpacking
print(item, price * quantity)
````

The next cell raises an error **on purpose**: a tuple cannot be changed.

````python
line[2] = 4
````

````python
order = [("Espresso", 1.10, 3), ("Croissant", 1.80, 1)]   # a list of tuples: a small table
print(order[1])
print(order[1][0])
````

### Dictionaries

A **dictionary** maps keys to values – like a menu maps items to prices.

````python
menu = {"Espresso": 1.10, "Croissant": 1.80, "Cappuccino": 2.40}
print(menu["Croissant"])
print(menu.get("Tea", 0.0))   # default value for a missing key
menu["Tea"] = 1.50            # add a new entry
print("Tea" in menu, len(menu))
print(menu)
````

A missing key raises a `KeyError` – **on purpose** in the next cell.

````python
menu["Hot Chocolate"]
````

### Sets

A **set** holds every value only once and has no order. `sorted()` shows it in a fixed order.

````python
student_items = {"Espresso", "Muffin", "Espresso"}
other_items = {"Espresso", "Pretzel"}
print(len(student_items))
print(sorted(student_items & other_items))   # bought by both
print(sorted(student_items | other_items))   # bought by anyone
print(sorted(student_items - other_items))   # only by students
````

### Shared references to mutable objects

An assignment never copies an object. Both names below refer to the **same** list.

````python
order = ["Espresso"]
tray = order
tray.append("Muffin")
print(order)
print(tray is order)
````

````python
tray = order.copy()   # an independent copy
tray.append("Tea")
print(order)
print(tray)
````

## 3.7 Loops

### for loops

`for` takes one element after the other and runs the indented body for each.

````python
order = [("Espresso", 3), ("Croissant", 1)]
menu = {"Espresso": 1.10, "Croissant": 1.80, "Cappuccino": 2.40}

for item, quantity in order:
    price = menu[item]
    print(f"{quantity} x {item}: {price * quantity:.2f}")
````

````python
for i in range(3):
    print(i)

for item, price in menu.items():
    print(item, price)
````

### Three loop patterns

````python
# 1 Accumulate: start with a neutral value and update it in every round
total = 0
for item, quantity in order:
    total += menu[item] * quantity
print(round(total, 2))
````

````python
# 2 Filter: keep only the elements that meet a condition
coffees = []
for item, quantity in order:
    if item in ("Espresso", "Cappuccino"):
        coffees.append(item)
print(coffees)
````

````python
# 3 Find the maximum: remember the best element so far
best, best_price = None, 0
for item, price in menu.items():
    if price > best_price:
        best, best_price = item, price
print(best)
````

Python has built-in functions for frequent patterns:

````python
print(max(menu, key=menu.get))   # the key with the largest value
print(sum(menu.values()))
````

### while, break and continue

Run the cell, type a few items and finish with `done`.

````python
total = 0
while True:
    item = input("Item (or 'done'): ")
    if item == "done":
        break
    if item not in menu:
        print("Unknown item")
        continue
    total += menu[item]

print(f"Total: {total:.2f} EUR")
````

### try and except

### From pseudocode to Python

The textbook by Russell and Norvig writes algorithms in pseudocode like this:

```
function RECEIPT-TOTAL(order, menu) returns a number
  total ← 0
  for each (item, quantity) in order do
    total ← total + menu[item] × quantity
  if total ≥ 20 then
    total ← total × 0.95
  return total
```

````python
def receipt_total(order, menu):
    total = 0
    for item, quantity in order:
        total = total + menu[item] * quantity
    if total >= 20:
        total = total * 0.95
    return total

print(round(receipt_total(order, menu), 2))
print(round(receipt_total([("Cappuccino", 9)], menu), 2))   # 21.60 EUR minus 5 %
````

## 3.8 Classes and Objects

A **class** is a blueprint: it defines the data each object stores (**attributes**) and what it can do (**methods**).

````python
class Receipt:
    """A receipt of the Campus Café."""

    def __init__(self, is_student=False):
        self.lines = []
        self.is_student = is_student

    def add(self, item, price, quantity=1):
        self.lines.append((item, price, quantity))

    def total(self):
        subtotal = 0
        for item, price, quantity in self.lines:
            subtotal += price * quantity
        if self.is_student:
            subtotal = subtotal * 0.9
        return round(subtotal, 2)
````

````python
anna = Receipt(is_student=True)
anna.add("Espresso", 1.10, 3)
anna.add("Croissant", 1.80)
print(anna.total())

ben = Receipt()
ben.add("Cappuccino", 2.40)
print(ben.total())
print(len(anna.lines), len(ben.lines))   # each object has its own data
````

### Asking questions with pandas

````python
sales["revenue"] = sales["quantity"] * sales["unit_price"]
sales["hour"] = sales["timestamp"].dt.hour

top = (sales.groupby("item")["quantity"].sum()
            .sort_values(ascending=False))
print(top.head(3))

food = sales[sales["category"] == "food"]
print(round(food["revenue"].sum(), 2))

per_receipt = sales.groupby("receipt_id")["revenue"].sum()
print(round(per_receipt.mean(), 2))
````

### matplotlib: revenue per hour

````python
import matplotlib.pyplot as plt

per_hour = sales.groupby("hour")["revenue"].sum()
per_hour.plot(kind="bar", rot=0)
plt.xlabel("Hour of the day")
plt.ylabel("Revenue in EUR")
plt.title("Campus Café, October 2025")
plt.show()
````
